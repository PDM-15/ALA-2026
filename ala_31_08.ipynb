```python
import numpy as np

a = np.array([1,2,3])  #--> len|shape|type|address of array

assert(a.dtype == np.int64)

assert(a.shape == (3,)) # shape returns a tuple always

#assert(a.shape != (3)) assert(a.shape!=3) --> both are treated as constants and not a tuple


assert isinstance(a.shape, tuple)
```

```python
a = np.zeros(4)

a == [0,0,0,0] # component wise comparison

np.array_equal(a, [1,2,3]) #compare two numpy objects
#.all() and .any() --> all compares all the elements in the array and gives the result; any compares atleast 1 to be present
```
